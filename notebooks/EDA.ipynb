```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('../data/Telco_customer_churn.csv')
```

```python
df.head(3)
```

```python
drop_cols = [
    "CustomerID",
    "Count",
    "Country",
    "State",
    "City",
    "Zip Code",
    "Lat Long",
    "Latitude",
    "Longitude",
    "Churn Label",
    "Churn Score",
    "CLTV",
    "Churn Reason"
]

df=df.drop(columns=drop_cols)
```

```python
df.head(3)
```

```python
categorical_cols = df.select_dtypes(include='object').columns
print("Categorical columns:", categorical_cols,"unique values:",df[categorical_cols].nunique())
```

```python
df["Total Charges"] = pd.to_numeric(df["Total Charges"], errors='coerce')
```

```python
df["Total Charges"] = df["Total Charges"].fillna(np.median(df["Total Charges"]))
```

```python
df.head()
```

```python
df.columns = df.columns.str.replace(" ", "_")
```

```python
df.info()
```

```python
categorical_cols = df.select_dtypes(include='object').columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
print("Categorical columns:", categorical_cols)
print("Numerical columns:", numerical_cols)
numerical_cols = numerical_cols.drop("Churn_Value")
```

```python
X=df.drop(columns=["Churn_Value"])
```

```python
y=df["Churn_Value"]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
cat_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor = ColumnTransformer([
    ("num", num_pipeline, numerical_cols),
    ("cat", cat_pipeline, categorical_cols)
])
```

```python
X_transformed = preprocessor.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_transformed, y, test_size=0.2, random_state=42,stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression
log_model = LogisticRegression()
log_model.fit(X_train, y_train)

```

```python
log_model.get_params()
```

```python
log_preds=log_model.predict(X_test)
```

```python
# Evaluate the model
from sklearn.metrics import accuracy_score,f1_score,recall_score,precision_score
accuracy = accuracy_score(y_test, log_preds)
f1 = f1_score(y_test, log_preds)
recall = recall_score(y_test, log_preds)
precision = precision_score(y_test, log_preds)

print(f"Accuracy: {accuracy:.4f}")
print(f"F1 Score: {f1:.4f}")
print(f"Recall: {recall:.4f}")
print(f"Precision: {precision:.4f}")
```

```python
import mlflow

mlflow.set_experiment("churn-prediction")
mlflow.set_tracking_uri("http://localhost:5000")
with mlflow.start_run(run_name="logistic_regression"):
    mlflow.log_param("model", "Logistic Regression")
    mlflow.log_metric("accuracy", accuracy)
    mlflow.log_metric("f1_score", f1)
    mlflow.log_metric("recall", recall)
    mlflow.log_metric("precision", precision)
    mlflow.sklearn.log_model(log_model, "model")
```

```python
from sklearn.ensemble import RandomForestClassifier
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)
```

```python
rf_preds=rf_model.predict(X_test)
```

```python
print(rf_preds)
```

```python
accuracy = accuracy_score(y_test, rf_preds)
print(f"Random Forest Accuracy: {accuracy:.4f}")

f1 = f1_score(y_test, rf_preds)
print(f"Random Forest F1 Score: {f1:.4f}")

recall = recall_score(y_test, rf_preds)
print(f"Random Forest Recall: {recall:.4f}")

precision = precision_score(y_test, rf_preds)
print(f"Random Forest Precision: {precision:.4f}")
```

```python
mlflow.set_experiment("churn-prediction")
mlflow.set_tracking_uri("http://localhost:5000")
with mlflow.start_run(run_name="random_forest"):
    mlflow.log_param("model", "Random Forest")
    mlflow.log_metric("accuracy", accuracy)
    mlflow.log_metric("f1_score", f1)
    mlflow.log_metric("recall", recall)
    mlflow.log_metric("precision", precision)
    mlflow.sklearn.log_model(rf_model, "model")
```

```python
# gradient boosting
from sklearn.ensemble import GradientBoostingClassifier
gb_model = GradientBoostingClassifier(random_state=42)
gb_model.fit(X_train, y_train)
```

```python
gd_preds=gb_model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, gd_preds)
print(f"Gradient Boosting Accuracy: {accuracy:.4f}")

f1 = f1_score(y_test, gd_preds)
print(f"Gradient Boosting F1 Score: {f1:.4f}")

recall = recall_score(y_test, gd_preds)
print(f"Gradient Boosting Recall: {recall:.4f}")
precision = precision_score(y_test, gd_preds)
print(f"Gradient Boosting Precision: {precision:.4f}")
```

```python
mlflow.set_experiment("churn-prediction")
mlflow.set_tracking_uri("http://localhost:5000")
with mlflow.start_run(run_name="gradient_boosting"):
    mlflow.log_param("gd_model", "Gradient Boosting")
    mlflow.log_metric("accuracy", accuracy)
    mlflow.log_metric("f1_score", f1)
    mlflow.log_metric("recall", recall)
    mlflow.log_metric("precision", precision)
    mlflow.sklearn.log_model(gb_model, "model")
```
